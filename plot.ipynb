```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("scores.csv")
# Remove rows where score is 0
df = df[df["score"] != 0]

fig, ax = plt.subplots(figsize=(8, 5))
ax.hist(
    df["score"],
    bins=range(int(df["score"].min()), int(df["score"].max()) + 2),
    edgecolor="black",
    align="left",
)
ax.set_xlabel("Score")
ax.set_ylabel("Count")
ax.set_title("Distribution of lottery scores (cumulative, unpenalized)")
ax.xaxis.set_major_locator(plt.MaxNLocator(integer=True))
plt.tight_layout()
plt.show()
```
